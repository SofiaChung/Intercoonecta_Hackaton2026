# Actividad 1: Exploración de datos de pingüinos

## Descripción del conjunto de datos

Este conjunto de datos contiene registros morfométricos de pingüinos del archipiélago Palmer, en la Antártida.

Incluye información sobre la especie, isla, dimensiones del pico, longitud de aleta, masa corporal, sexo y año.

Fuente abierta: Palmer Penguins. https://zenodo.org/records/12772944

## Objetivo

Importar un archivo CSV utilizando R y explorar sus datos.

```r
# Dirección del conjunto de datos de de base abierta
url <- "https://zenodo.org/records/12772944/files/pinguinos.csv?download=1"
```

```r
# Leer el archivo CSV directamente desde internet
pinguinos <- read.csv(url)
```

```r
# Mostrar las primeras 6 filas
head(pinguinos)
```

## Actividad 2: Visualización y funciones en R

Se elaboraron tres gráficos exploratorios: una gráfico de barras para determinar la cantidad de registros por especie, un diagrama de cajas para comparar la masa corporal entre especies y un gráfico de dispersión para examinar la relación entre la longitud de aleta y la masa corporal.

Además, se definió una función básica que convierte el kilogramos en miligramos.

Estos procedimientos permiten explorar diferencias entre especies y practicar la creación de funciones reutilizables en R.

```r
# Gráfico 1: Contar cuántos registros hay por especie
conteo <- table(pinguinos$especie)

# Crear gráfico de barras
barplot(conteo,
        main = "Número de pingüinos por especie",
        xlab = "Especie",
        ylab = "Número de registros",
        col = "lightblue")
```

```r
# Gráfico 2: masa corporal por especie
boxplot(masa_corporal_g ~ especie,
        data = pinguinos,
        main = "Masa corporal por especie",
        xlab = "Especie",
        ylab = "Masa corporal (g)")
```

```r
# Gráfico 3: relación entre aleta y masa corporal
plot(pinguinos$largo_aleta_mm,
     pinguinos$masa_corporal_g,
     main = "Longitud de aleta y masa corporal",
     xlab = "Longitud de aleta (mm)",
     ylab = "Masa corporal (g)")
```

```r
# Crear una función que convierta gramos a kilogramos
gramos_a_kg <- function(gramos) {
  gramos / 1000
}

# Probar nuestra función
gramos_a_kg(3750)
```
