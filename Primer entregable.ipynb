# Actividad 1: Exploración de datos de pingüinos

## Descripción del conjunto de datos

Este conjunto de datos contiene registros morfométricos de pingüinos del archipiélago Palmer, en la Antártida.

Incluye información sobre la especie, isla, dimensiones del pico, longitud de aleta, masa corporal, sexo y año.

Fuente abierta: Palmer Penguins. https://zenodo.org/records/12772944

## Objetivo

Importar un archivo CSV utilizando R y explorar sus datos.

```r
# Dirección del conjunto de datos de de base abierta
url <- "https://zenodo.org/records/12772944/files/pinguinos.csv?download=1"
```

```r
# Leer el archivo CSV directamente desde internet
pinguinos <- read.csv(url)
```
